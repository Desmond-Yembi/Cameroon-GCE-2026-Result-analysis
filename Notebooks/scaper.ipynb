```python
import pandas as pd
import os
```

### Scraping 2026 results

```python
# --------------------------------------------------
# 1. List of GCE Board URLs
# --------------------------------------------------

urls = [
    "https://camgceb.org/results-statistics/overall-performance/",
    "https://camgceb.org/results-statistics/global-performance-by-region/",
    "https://camgceb.org/results-statistics/performance-by-school-type/",
    "https://camgceb.org/results-statistics/results-by-regions/",
    "https://camgceb.org/results-statistics/results-by-divisions/",
    "https://camgceb.org/results-statistics/performance-by-subjects/",
    "https://camgceb.org/results-statistics/performance-by-specialty/",
    "https://camgceb.org/results-statistics/performance-by-specialty/",
    "https://camgceb.org/results-statistics/malpractices/",
    "https://camgceb.org/results-statistics/overall-performance/",
    "https://camgceb.org/results-statistics/honors-roll-candidates/",
    "https://camgceb.org/results-statistics/tvee-candidates-best-performances/",
    "https://camgceb.org/results-statistics/exam-centres-honors-roll-schools/"
    "https://camgceb.org/results-statistics/tvee-exam-centres-best-performances/",
    "https://camgceb.org/national-manpower-development/",
    



]

# --------------------------------------------------
# 2. Create main output folder
# --------------------------------------------------

output_folder = r"C:\Users\Desmond Yembi\Desktop\GCE_Results Analysis\data"

os.makedirs(output_folder, exist_ok=True)

# --------------------------------------------------
# 3. Loop through each URL
# --------------------------------------------------

for url in urls:

    print("\n" + "=" * 70)
    print("Processing:")
    print(url)
    print("=" * 70)

    try:

        # Extract all tables from the webpage
        tables = pd.read_html(url)

        print(f"Tables found: {len(tables)}")

        # --------------------------------------------------
        # 4. Create a folder based on the webpage
        # --------------------------------------------------

        page_name = url.rstrip("/").split("/")[-1]

        page_folder = os.path.join(output_folder, page_name)

        os.makedirs(page_folder, exist_ok=True)

        # --------------------------------------------------
        # 5. Loop through all tables
        # --------------------------------------------------

        for i, df in enumerate(tables):

            file_name = f"{page_name}_table_{i + 1}.csv"

            file_path = os.path.join(page_folder, file_name)

            # Save table
            df.to_csv(
                file_path,
                index=False,
                encoding="utf-8-sig"
            )

            print(f"Saved: {file_path}")

    except Exception as e:

        print(f"ERROR processing {url}")
        print(e)
```
